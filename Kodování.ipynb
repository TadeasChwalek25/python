# Kódování, binární řetězce a šifrování

```python
# 1) Převod textu na binární řetězec a hexadecimální zápis
text = "Hello world"
encoded_bytes = text.encode("utf-8")
print("Bajty:", list(encoded_bytes))
print("Hex zápis:", encoded_bytes.hex())
binary_string = ''.join(format(byte, '08b') for byte in encoded_bytes)
print("Binární řetězec:", binary_string)
```

Text je pomocí encode() převeden na bajty. 
Hex zobrazí hexadecimální tvar. 
Každý bajt můžu převést na 8bitový binární zápis.

```python
# 2) XOR šifrování a dešifrování
def xor_cipher(data: bytes, key: bytes) -> bytes:
    result = bytearray()
    for i in range(len(data)):
        result.append(data[i] ^ key[i % len(key)])
    return bytes(result)

original_text = "Secret message"
key = b"mobile"
original_bytes = original_text.encode("utf-8")
encrypted = xor_cipher(original_bytes, key)
print("Zašifrované (hex):", encrypted.hex())
decrypted = xor_cipher(encrypted, key)
print("Dešifrovaný text:", decrypted.decode("utf-8"))
```

XOR porovnává jednotlivé bity. 
Operace je symetrická, proto stejná funkce slouží pro šifrování i k dešifrování.

```python
# 3) Změna kódování souboru (UTF-8 -> UTF-16 LE)
input_file = "vstup_utf8.txt"
output_file = "vystup_utf16le.txt"

with open(input_file, "w", encoding="utf-8") as f:
    f.write("Ukázkový text s diakritikou: čřžýáíé")

with open(input_file, "r", encoding="utf-8") as f:
    content = f.read()

with open(output_file, "w", encoding="utf-16le") as f:
    f.write(content)

print("Soubor byl převeden z UTF-8 na UTF-16 LE.")
```

Soubor se načte v původním kódování a zapíše v novém. 
Python při zápisu automaticky provede překódování.
